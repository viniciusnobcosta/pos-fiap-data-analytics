### 1. Entendimento do Negócio

**1.1. Contexto de negócio**

O projeto tem como contexto o setor de e-commerce brasileiro e utiliza o Brazilian E-Commerce Public Dataset by Olist, disponibilizado para a realização de análises sobre o comportamento e o desempenho de operações comerciais online. **1.4. Objetivo Analítico**

**1.5. Direcionamento Inicial**
